````python
import seaborn as sns
import matplotlib.pyplot as plt

diamonds = sns.load_dataset("diamonds")

plt.scatter(diamonds["carat"], diamonds["price"])
plt.xlabel("Carat")
plt.ylabel("Price")
plt.title("Carat vs Price")
plt.show()

plt.hist(diamonds["price"])
plt.xlabel("Price")
plt.ylabel("Count")
plt.title("Price Histogram")
plt.show()

sns.boxplot(x="cut", y="price", data=diamonds)
plt.xlabel("Cut")
plt.ylabel("Price")
plt.title("Cut vs Price")
plt.show()


````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
import seaborn as sns

diamonds = sns.load_dataset("diamonds")
def convert(c):
    g = c * 0.2
    return g
    
heaviest_row = diamonds.loc[diamonds["carat"].idxmax()]
heaviest_grams = convert(heaviest_row["carat"])

print(heaviest_grams)



````
